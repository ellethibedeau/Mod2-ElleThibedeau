# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

## Answer: 

First visual: 
Advantages - having two plots side by side shows a clear separation between the two weeks. Easier to see trends for a specific week without clutter or overlap, useful when studying weekly patterns. 
Disadvantages - Harder to directly compare the two weeks, space consuming

Second visual:
Advantages - easy to compare the two weeks directly on a shared axis, saves space, clear visual to see overlaps and patterns between the two weeks
Disadvantages - Cluttered or harder to read if there is a lot of overlap

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python


departments = ['HR', 'Finance', 'Engineering', 'Sales', 'Marketing']
scores = [3.8, 4.2, 4.5, 3.9, 4.1]


departments_sorted = [d for _, d in sorted(zip(scores, departments))]
scores_sorted = sorted(scores)

plt.figure(figsize=(8, 5))
plt.hlines(y=departments_sorted, xmin=0, xmax=scores_sorted, color='gray', linestyles='dotted')
plt.plot(scores_sorted, departments_sorted, 'o', color='steelblue')


for i, score in enumerate(scores_sorted):
    plt.text(score + 0.05, departments_sorted[i], f'{score:.1f}', va='center', fontsize=9)

plt.title('Average Satisfaction Score by Department', fontsize=14)
plt.xlabel('Score (out of 5)')
plt.xlim(0, 5)
plt.grid(False)
plt.tight_layout()
plt.show()
```
